```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('train.csv',usecols=['Cabin','Ticket','Survived'])
```

```python
df.head()
```

## code for one have diffrent numbers and leeters 

```python
# extract numerical part
df['Numerical_column']=pd.to_numeric(df['Cabin'],errors='coerce',downcast='integer')
```

```python
# extract categorocal part
df['categoracal_column']=np.where(df['Numerical_column'].isnull(),df['Cabin'],np.nan)

df.head()
```

## 2nd technique for the concarente values

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
df['Cabin_num']=df['Cabin'].str.extract('(/d+)') #capture the numerical part
df['Cabin_cat']=df['Cabin'].str[0]
df.head()
```

```python
df['Cabin_cat'].value_counts().plot(kind='bar')
```

## Handle Ticket columns 

```python
df['Ticket_num']=df['Ticket'].apply(lambda s: s.split()[-1])
df['Ticket_num']=pd.to_numeric(df['Ticket_num'],errors='coerce',downcast='integer')

#extract first pRT OF TICKET AS CATEGORY
df['Ticket_cat']=df['Ticket'].apply(lambda s: s.split()[0])
df['Ticket_cat']=np.where(df['Ticket_cat'].str.isdigit(),np.nan,df['Ticket_cat'])

df.head()
```
